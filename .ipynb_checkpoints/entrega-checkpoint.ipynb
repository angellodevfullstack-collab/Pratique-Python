# Leitura da base do ecommerce 🛒

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurações de estilo
sns.set(style="whitegrid", context="talk")
plt.rcParams["figure.figsize"] = (12, 7)

# Substitua pelo caminho do seu arquivo
csv_path = "ecommerce_estatistica.csv"

# Leitura do CSV
df = pd.read_csv(csv_path)

# Visualização inicial
display(df.head())
print("\nInfo:")
print(df.info())
print("\nEstatísticas descritivas (numéricas):")
display(df.describe(include=[np.number]))
print("\nEstatísticas descritivas (categóricas):")
display(df.describe(include='object'))
```

# Limpezas e Tratativas Iniciais 🧹

```python
# Remover coluna índice redundante
if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])

# Padronizar nomes de colunas (sem espaços, minúsculas)
df.columns = (
    df.columns.str.strip()
              .str.lower()
              .str.replace(" ", "_")
              .str.replace("/", "_")
              .str.replace("-", "_")
)

# Conversões de tipos
# Preço
df["preço"] = pd.to_numeric(df["preço"], errors="coerce")

# Nota e N_Avaliações
df["nota"] = pd.to_numeric(df["nota"], errors="coerce")
df["n_avaliações"] = pd.to_numeric(df["n_avaliações"], errors="coerce")

# Desconto (algumas linhas podem estar vazias)
df["desconto"] = pd.to_numeric(df["desconto"], errors="coerce")

# Qtd_Vendidos: trata valores como "+10mil" e "1000"
def parse_qtd_vendidos(x):
    if pd.isna(x):
        return np.nan
    x = str(x).strip().lower().replace(".", "").replace(",", ".")
    if "mil" in x:
        # tenta captar números antes de "mil"
        # exemplos: "+10mil", "10 mil", "mais de 10 mil"
        num = "".join([ch for ch in x if ch.isdigit() or ch == "."])
        try:
            return float(num) * 1000
        except:
            return 10000.0  # fallback razoável
    # remove sinais de "+", etc.
    x = x.replace("+", "")
    try:
        return float(x)
    except:
        return np.nan

df["qtd_vendidos"] = df["qtd_vendidos"].apply(parse_qtd_vendidos)

# Correções de codificados (se existirem)
for col in ["marca_cod","material_cod","temporada_cod","qtd_vendidos_cod",
            "nota_minmax","n_avaliações_minmax","desconto_minmax","preço_minmax",
            "marca_freq","material_freq"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Categóricas principais
for col in ["marca", "material", "gênero", "temporada", "título"]:
    if col in df.columns:
        df[col] = df[col].astype("category")

# Remover duplicatas por título (se necessário)
df = df.drop_duplicates(subset=["título"], keep="first")

# Remover linhas com preço ou nota ausentes para análises chave
df_clean = df.copy()
df_clean = df_clean.dropna(subset=["preço", "nota"])

print("\nApós limpeza:")
print(df_clean.info())
display(df_clean.head())
```

# Gráfico de barra 📊 (Top marcas por número de produtos)

```python
# Agrupar vendas por marca e selecionar Top 10
top_brands = (
    df_clean.groupby("marca")["qtd_vendidos"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

# Gráfico de barras horizontais
plt.figure(figsize=(10,6))
top_brands.plot(kind='barh')
plt.title("Top 10 Marcas por Volume de Vendas")
plt.xlabel("Quantidade Vendida")
plt.ylabel("Marca")
plt.tight_layout()
plt.show()
```

# Gráfico de pizza 🍕 - Top 10 Materiais por Volume de Vendas

```python
# Agrupar vendas por material e selecionar Top 10
top_materials = (
    df_clean.groupby("material")["qtd_vendidos"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

# Plotar gráfico de pizza com % mais afastadas do centro
plt.figure(figsize=(8,8))
wedges, texts, autotexts = plt.pie(
    top_materials,
    labels=None,
    autopct="%1.1f%%",
    pctdistance=0.9,     # 🔹 Aumenta a distância das porcentagens
    startangle=90
)

# Ajustar tamanho da fonte das porcentagens
for autotext in autotexts:
    autotext.set_fontsize(8)

# Legenda separada
plt.legend(
    wedges,
    top_materials.index,
    title="Materiais",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)

plt.title("Top 10 Materiais por Volume de Vendas")
plt.tight_layout()
plt.show()
```

# Gráfico de densidade 📈 (KDE de preço por material)

```python
# Garantir que preço é numérico
df_clean["preço"] = pd.to_numeric(df_clean["preço"], errors="coerce")

# Remover linhas sem preço ou material
df_clean_kde = df_clean.dropna(subset=["preço", "material"])

top5_materials = (
    df_clean_kde.groupby("material")["qtd_vendidos_cod"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
    .index
)

df_top5 = df_clean_kde[df_clean_kde["material"].isin(top5_materials)]

plt.figure(figsize=(10,6))

for material in top5_materials:
    subset = df_top5[df_top5["material"] == material]
    sns.kdeplot(
        data=subset,
        x="preço", 
        fill=True,
        alpha=0.3,
        linewidth=1,
        label=material
    )

plt.title("Distribuição de Preços dos Top 5 Materiais")
plt.xlabel("Preço (R$)")
plt.ylabel("Densidade")
plt.legend(title="Material", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
